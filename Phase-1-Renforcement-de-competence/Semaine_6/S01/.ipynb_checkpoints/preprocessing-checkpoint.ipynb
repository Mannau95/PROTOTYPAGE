```python
import pandas as pd
import numpy as np
import os

# Charger les 1000 transactions
chemin = r"D:\Nouveau dossier\Prototypage\transactions.jsonl"
df = pd.read_json(chemin, lines=True)
print(f"Nombre de transactions : {len(df)}")
df.head()
```

```python
# Convertir timestamp en datetime
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Extraire heure et jour de la semaine
df['heure'] = df['timestamp'].dt.hour
df['jour_semaine'] = df['timestamp'].dt.dayofweek  # 0=lundi, 6=dimanche

# Encoder le canal en variables binaires (one-hot)
df_canal = pd.get_dummies(df['canal'], prefix='canal')

# Récupérer les coordonnées géolocalisation
df['lat'] = df['geolocalisation'].apply(lambda x: x['lat'] if isinstance(x, dict) else None)
df['lon'] = df['geolocalisation'].apply(lambda x: x['lon'] if isinstance(x, dict) else None)


# Sélectionner les colonnes de base qui sont dans df
features_base = ['montant', 'heure', 'jour_semaine', 'duree_session_ms', 
                 'code_reponse', 'lat', 'lon']

# Concaténer les colonnes de base avec les colonnes one-hot (df_canal)
X = pd.concat([df[features_base], df_canal], axis=1)

# Remplacer les valeurs manquantes
X = X.fillna(0)

print("Shape de X :", X.shape)
X.head()
```

```python
# Créer y : 1 si transaction suspecte/erreur, 0 sinon
df['is_anomaly'] = df['statut'].isin(['suspect', 'erreur']).astype(int)
y = df['is_anomaly']

print("Proportion d'anomalies :", y.mean())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=42, stratify=y
)

print("Train :", X_train.shape)
print("Test :", X_test.shape)
```

```python
print("✅ Données chargées :", len(df))
print("✅ Features :", X_train.shape[1])
print("✅ Anomalies dans train :", y_train.sum())
print("✅ Anomalies dans test :", y_test.sum())
print("✅ Proportion train :", y_train.mean())
```
