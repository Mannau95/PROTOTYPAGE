```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Charger le fichier JSON Lines
df = pd.read_json('transactions.jsonl', lines=True)

# Afficher les 5 premières lignes
df.head()
# Informations générales (types, non-null, mémoire)
df.info()
# Statistiques descriptives pour les colonnes numériques
df.describe()
# Statistiques incluant les colonnes catégorielles
df.describe(include='all')
# Vérifier les types actuels
df.dtypes
# Vérifier les doublons sur l'id
print(f"Doublons sur 'id': {df['id'].duplicated().sum()}")
# Supprimer les doublons
df = df.drop_duplicates(subset='id')
# Vérifier les montants négatifs ou nuls
print(f"Montants <= 0: {(df['montant'] <= 0).sum()}")

# Convertir 'timestamp' en datetime
df['timestamp'] = pd.to_datetime(df['timestamp'])

# S'assurer que 'montant' est numérique (déjà le cas)
df['montant'] = df['montant'].astype(float)

# Vérifier les valeurs manquantes
df.isnull().sum()


# Filtrer (si nécessaire)
df = df[df['montant'] > 0]
df.to_csv('transactions_nettoyees.csv', index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('transactions_nettoyees.csv', parse_dates=['timestamp'])
# Afficher les colonnes du DataFrame
print(df.columns.tolist())
```

```python
# Créer les colonnes dérivées une fois pour toutes
df['timestamp'] = pd.to_datetime(df['timestamp'])   # si pas déjà fait
df['heure'] = df['timestamp'].dt.hour
df['jour_semaine'] = df['timestamp'].dt.day_name()
```

```python
# Vérifier que les colonnes ont bien été créées
print(df[['timestamp', 'heure', 'jour_semaine']].head())
```

```python
# Statistiques par statut
stats_statut = df['statut'].value_counts()
print(stats_statut)

stats_statut_percent = df['statut'].value_counts(normalize=True) * 100
print(stats_statut_percent)

```

```python
# Activité par heure
activite_heure = df.groupby('heure').size()
print(activite_heure)

# Visualisation
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 5))
activite_heure.plot(kind='line', marker='o')
plt.title('Nombre de transactions par heure')
plt.xlabel('Heure')
plt.ylabel('Nombre de transactions')
plt.grid(True)
plt.show()
```

```python
# Créer les colonnes dérivées une fois pour toutes
df['heure'] = df['timestamp'].dt.hour
suspects = df[df['statut'] == 'suspect']

# Répartition par canal
suspects_par_canal = suspects['canal'].value_counts()
print("Suspects par canal :\n", suspects_par_canal)

# Répartition par heure
suspects_par_heure = suspects.groupby('heure').size()
print("Suspects par heure :\n", suspects_par_heure)

# Montant moyen des suspects par canal
montant_moyen_suspect_par_canal = suspects.groupby('canal')['montant'].mean()
print("Montant moyen suspect par canal :\n", montant_moyen_suspect_par_canal)
```

```python
# Compter le nombre total de transactions par jour
total_par_jour = df.groupby(df['timestamp'].dt.date).size()

# Compter les transactions suspectes par jour
suspect_par_jour = df[df['statut'] == 'suspect'].groupby(df['timestamp'].dt.date).size()

# Taux d'anomalie
taux_anomalie = (suspect_par_jour / total_par_jour).fillna(0)
print("Taux d'anomalie par jour :\n", taux_anomalie)

# Tracer le taux d'anomalie dans le temps
plt.figure(figsize=(12, 4))
taux_anomalie.plot(kind='line', marker='o')
plt.title('Taux de transactions suspectes par jour')
plt.xlabel('Date')
plt.ylabel('Taux d\'anomalie')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Style Seaborn pour des graphiques plus agréables
sns.set_style("whitegrid")
sns.set_palette("Set2")

# Pour afficher les graphiques dans le notebook
%matplotlib inline
```

```python
plt.figure(figsize=(12, 6))

# Histogramme des montants normaux (statut != 'suspect')
df_normaux = df[df['statut'] != 'suspect']
plt.hist(df_normaux['montant'], bins=50, alpha=0.7, label='Normaux', color='blue')

# Histogramme des montants suspects
df_suspects = df[df['statut'] == 'suspect']
plt.hist(df_suspects['montant'], bins=50, alpha=0.7, label='Suspects', color='red')

plt.title('Distribution des montants des transactions')
plt.xlabel('Montant (FCFA)')
plt.ylabel('Nombre de transactions')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='canal', y='montant', data=df, palette='Set2')
plt.title('Distribution des montants par canal')
plt.xlabel('Canal')
plt.ylabel('Montant (FCFA)')

plt.show()
```

```python
# Agrégation du nombre de transactions par heure
activite_heure = df.groupby('heure').size()

plt.figure(figsize=(12, 5))
activite_heure.plot(kind='line', marker='o', color='teal', linewidth=2)
plt.title('Nombre de transactions par heure de la journée')
plt.xlabel('Heure')
plt.ylabel('Nombre de transactions')
plt.xticks(range(0, 24))
plt.grid(True, alpha=0.3)

# Ligne verticale pour les horaires de bureau (8h-18h)
plt.axvspan(8, 18, alpha=0.2, color='gray', label='Horaires de bureau')
plt.legend()
plt.show()
```
